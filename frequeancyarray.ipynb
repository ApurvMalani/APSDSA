```c++
// Using  Map Counting  Frequncy of  Array 
#include <bits/stdc++.h>
using namespace std;

int main()
{
	int n;
	cin >> n;

	int arr[n];

	for (int i = 0; i < n; i++)
	{
		cin >> arr[i];
	}

	map<int, int> mpp;

	for (int i = 0; i < n; i++)
	{
		mpp[arr[i]]++;
	}

	int nn;
	cin >> nn;

	while (nn--)
	{
		int numbers;
		cin >> numbers;

		cout << mpp[numbers] << endl;
	}

	return 0;
}
```

```c++
//  Using Hash.. 

#include <bits/stdc++.h>
using namespace std;

int main()
{
	int n;
	cin >> n;

	int arr[n];

	for (int i = 0; i < n; ++i)
	{
		cin >> arr[i];
	}

	int hash[13] = {0};
	for (int i = 0; i < n; ++i)
	{
		hash[arr[i]] += 1;
	}

	int h;
	cin >> h;

	while (h--)
	{
		int numbers;
		cin >> numbers;

		cout << hash[numbers] << " " << endl;
	}

	return 0;
}
```
